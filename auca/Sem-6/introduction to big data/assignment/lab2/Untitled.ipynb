```python
with open("week3_students.csv") as f:
    lines = f.readlines()
    print(len(lines))
    print(lines[0])
    print(lines[1])
        
    
```

```python
names=[]
count=0
for line in lines[1:]:
    row=line.strip().split(",")
    name = row[1]
    names.append(name)
    count +=1

for name in names[0:5]:
    print(name)

print(count)

    
```

```python
count=0
total = 0
valid = 0
bad = 0
bad_ids=[]

for line in lines[1:]:
    parts=line.strip().split(",")
    score=parts[7]
    try:
        total += int(score)
        valid +=1
        count +=1
    except ValueError:
        bad +=1
        bad_id=parts[0]
        bad_ids.append(bad_id)

print(total)
print(f"{total/count:.1f}")
print(valid)
print(bad)
for id in bad_ids:
    print(id)
```

```python
passed=0
failed = 0
best_score =0
best_name=""

for line in lines[1:]:
    parts=line.strip().split(",")
   
    try:
        score = int(parts[7])
        if score > best_score:
            best_score = score
            best_name = parts[1]
        if score >=50:
            passed +=1
        else:
            failed +=1
    except ValueError:
        pass

print(passed)
print(failed)
print(best_score)
print(best_name)
```

```python
sums={}
counts={}

for line in lines[1:]:
    parts=line.strip().split(",")

    try:
        score = int(parts[7])
        d=parts[5]
        sums[d] = sums.get(d, 0) + score
        counts[d] = counts.get(d, 0) + 1
    except ValueError:
        pass

with open("report.txt" , "w") as f:
        
    for d in sums:
        print (f"{d}: {sums[d]/counts[d]:.1f}\n")
        f.write(f"{d}: {sums[d]/counts[d]:.1f}\n")
        
```
