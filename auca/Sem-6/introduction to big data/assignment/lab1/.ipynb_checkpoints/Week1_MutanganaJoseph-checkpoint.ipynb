## Part 6 — Reflection

### 1. Is this dataset Big Data? Does it demonstrate the 5 V's?

No, this dataset is not Big Data because it contains information for only 20 students, so its volume is very small.

**V's the dataset  doesn't cleary have**

* **Volume:** The dataset does not have a large volume because it contains only 20 student records.
* **Variety:** The dataset contains different data types such as strings, integers, and Boolean-like values, but it is still mainly one structured dataset. It does not contain different forms of data such as images, videos, audio, or JSON.
* **Velocity:** The dataset does not demonstrate high velocity because the data is already collected. It is not being generated or processed continuously at high speed.
* **Veracity:** We cannot fully verify the accuracy and reliability of the information from the dataset alone. For example, we cannot independently confirm whether every record really belongs to an AUCA student or whether every score and attendance value is accurate.


Overall, the dataset is useful for learning Python and basic data analysis, but it is too small and simple to demonstrate the challenges of Big Data.

### 2. What surprised or confused me?

One thing that surprised me was how much useful information could be obtained from a small dataset using simple Python functions such as `max()`, `min()`, `sum()`, and `len()`. This helped me understand how Python can be used as a foundation for more advanced data analysis and Big Data tools.
