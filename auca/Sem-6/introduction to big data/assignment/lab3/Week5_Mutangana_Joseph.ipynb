# Load data into arrays



```python
import numpy as np

FILE = "week5_kigali_weather.csv"

date = np.loadtxt(FILE,delimiter=",", skiprows=1, usecols=1, dtype=str)
temps,rain,hum = np.genfromtxt(FILE, delimiter=",", skip_header=1, usecols=(2,3,4), unpack=True)
print(f"Total {len(date)} Days")
```

# Part 1 · List Comprehensions

```python
temps_list = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1] # given for Part 1
raw = ["24.4", "25.9", "28.6", "19.3"] # given for Ex 1.3
```

## Exercise 1.1 — Transform

```python
convert = [(t * 9/5 + 32) for t in temps_list]
print(convert)
```

## Exercise 1.2 — Filter

```python
hot = [t for t in temps_list if t >= 26]
cool_rounded = [round(t) for t in temps_list if t<24]

print(f"Hot: {hot}")
print(f"Cool_rounded: {cool_rounded}")
```

## Exercise 1.3 — Convert types

```python
new_row = [float(r) for r in raw]
average=sum(new_row)/len(new_row)
print(f"Average: {average}")
```

# Part 2 · lambda

## Exercise 2.1 — Rank with sorted


```python
results = [("Aline", 85), ("Eric", 78), ("Diane", 91), ("Samuel", 92), ("Grace",76)]

sort_score=sorted(results, key=lambda s:s[1], reverse=True)
print(sort_score[0:3])
print()
print(f"1st: {sort_score[0][0]} - {sort_score[0][1]}")
print(f"2nd: {sort_score[1][0]} - {sort_score[1][1]}")
print(f"3rd: {sort_score[2][0]} - {sort_score[2][1]}")
```

## Exercise 2.2 — max with a key

```python
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest=max(rain_log, key= lambda r:r[1])
print(f"Wettest day is  {wettest[0]}, with rain fall of {wettest[1]} mm.")
```

## Exercise 3.2 — Describe the temperatures 

```python
print(f"The mean is: {temps.mean():.2f} °C")
print(f"The max is: {temps.max():.2f} °C")
print(f"The min is: {temps.min():.2f} °C")
print(f"The std is: {temps.std():.2f} °C")
```

## Exercise 3.3 — The hottest day

```python
position = np.argmax(temps)
hottest_day = date[position]
temperature = temps[position]
print(f"Hottest day is {hottest_day} at {temperature} °C")
```

# Part 4 · Boolean Masking

## Exercise 4.1 — Hot days

```python
hot_days = temps >=26
print(f"Hot days are: {hot_days.sum()}")
print(f"Average of hot days: {temps[hot_days].mean():.2f}")
```

## Exercise 4.2 — Rain report`m

```python
rain_mask = rain >0
print(f"Number of rain days: {rain_mask.sum()}")
print(f"Total rainfall over 60 days: {rain.sum():.2f} mm")
pos=np.argmax(rain)
print(f"Heviest single day rainfall: {date[pos]} with {rain[pos]} mm")
```

## Exercise 4.3 — Cross-array analysis

```python
print(f"Average temps on rainy day: {temps[rain_mask].mean():.2f}")
print(f"Average temps on dry day: {temps[~rain_mask].mean():.2f}")
print(f"Hot and rainy day: {((temps >=26) & (rain>0)).sum()}")
print(f"Days with humidity >= 70: {(hum>=70).sum()}")
```

# Part 5 · Anomalies & Trends 

## Exercise 5.1 — Anomaly detection

```python
anomaly = temps-temps.mean()
mask = anomaly > 3
print(f"Dates bove 3 °C on temperature mean: {date[mask]}")
print(f"Days 3 °C below the temperature mean: {(anomaly < -  3).sum()}")
```

## Exercise 5.2 — Is the season changing?

```python
print(f"Average temperaturer of 30 first days: {temps[:30].mean():.2f}")
print(f"Average temperaturer of 30 last days: {temps[30:].mean():.2f}")
```
