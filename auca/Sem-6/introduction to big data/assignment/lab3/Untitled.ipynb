```python

import numpy as np

file = "week5_kigali_weather.csv"

date = np.loadtxt(file, delimiter=",", skiprows=1, usecols=1, dtype=str)
temps,rainfall,humidity = np.genfromtxt(file,delimiter=",",skip_header=1, usecols=(2,3,4), unpack=True)

```

```python
templ = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1] # given for Part 1
raw = ["24.4", "25.9", "28.6", "19.3"]

f = [t * 9/5 + 35 for t in templ]
hot=[t for t in templ if t >= 26]
cool_rounded=[round(t) for t in templ if t <24]
              

print(f"Fahrenheit: {f}")
print(f"Hot: {hot}")
print(f"Cool rounded: {cool_rounded}")
```

```python
print(f"Mean: {temps.mean():.2f} C")
print(f"Max: {temps.max():.2f} C")
print(f"Min: {temps.min():.2f} C")
print(f"Std: {temps.std():.2f} C")
```

```python
index = np.argmax(temps)

print(f"The hostest day is {date[index]} with temperature of {temps[index]}")
```

```python
hot_days = temps >= 26

print(f"Hot days are: {hot_days.sum()}")
print(f"Average of hot days: {temps[hot_days].mean():.2f}")
```

```python
rain_msk = rainfall >0
print(f"Rainy days are: {rain_msk.sum()}")
print(f"Total rain fall over 60 days: {rainfall.sum():.2f}mm")
index = np.argmax(rainfall)
print(f"Heavy rain day: {date[index]} with {rainfall.max():.2f}mm")
```

```python
anomaly = temps - temps.mean()

humidity_days = humidity >=70
above_mean = anomaly > 3
below_mean = anomaly < - 3
print(date[above_mean])
print(below_mean.sum())
print(humidity_days.sum())
print(f"Hot and rain days: {((temps >=26) & (rainfall > 0)).sum()}")
```
